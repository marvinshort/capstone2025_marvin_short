```python
import pandas as pd
import numpy as np
import requests
import matplotlib.pyplot as plt
import seaborn as sns
import json
```

```python
file_path = "../data/keys.json"

with open(file_path, 'r') as file:
    credentials = json.load(file)
    
API_KEY = credentials['api_key']
```

##2015 DATA

```python
params = {'api_key' : API_KEY}

baseurl = 'https://api.eia.gov/v2/'

endpoint = 'electricity/retail-sales'

data_end_tag = '/data?'

facets = 'frequency=monthly&data[0]=customers&data[1]=sales&facets[sectorid][]=COM&facets[sectorid][]=IND&facets[sectorid][]=OTH&facets[sectorid][]=RES&facets[sectorid][]=TRA&start=2015-01&end=2015-12&sort[0][column]=period&sort[0][direction]=desc&offset=0&length=5000'

datapoint = '&data[]=value'

url = baseurl + endpoint + data_end_tag + facets

response = requests.get(url=url, params=params)

print(response.status_code)
print(url)
```

```python
data = response.json()
response.json()

#Write data to a dataframe for analysis
customers_2015 = pd.DataFrame(data['response']['data'])
customers_2015.shape
```
